# RQ3: дообучение mmBERT-small — CE, SupCon и иерархический SupCon

**Вопрос.** Улучшает ли дообучение разделимость safe/harm по сравнению с frozen-энкодером и что даёт
контрастивный сигнал — по классу и с сохранением структуры категорий вреда. Пять вариантов отличаются
только loss: данные, сэмплер батчей, инициализация и seed у них одинаковы, а общие гиперпараметры
(`SHARED_HPARAMS`, прежде всего `encoder_lr`) подбираются один раз на все варианты. Задача бинарная:
harm vs safe.

| вариант | loss | что проверяет |
|---|---|---|
| `ce` | cross-entropy линейной головы | базовое дообучение классификатора |
| `supcon` | SupCon по классу harm/safe | даёт ли контрастив по классу лучшее пространство, чем CE |
| `ce+supcon` | CE + λ · SupCon | помогает ли контрастив по классу обычному обучению |
| `hier-supcon` | взвешенный SupCon: пара → категория вреда → класс | помогает ли сохранение категорий вреда (против `supcon`) |
| `ce+hier-supcon` | CE + λ · взвешенный SupCon | помогает ли такой контрастив обучению классификатора (против `ce` и `ce+supcon`) |

`frozen` — mmBERT-small без обучения

**Данные** (подробно — `training_pairs_v2.md` в корне репозитория):

* **train** — пары v2 из `training_pairs` (AEGIS 2.0 train, обёртки HarmBench; code bank — пока не доабвляю, код домен будем рассматривать отдельно в дальшнейших экспериментах);
* **val** — AEGIS 2.0 `validation`: подбор гиперпараметров, C и k скореров, порогов; мониторинг геометрии;
* **hold-out** (in-distribution) — AEGIS 2.0 `test`;
* **OOD** (LODO) — два независимых источника: сэмпл ToxicChat (`toxicchat0124`, test) и WildGuardMix
  (по `WILDGUARD_SAMPLE_PER_CELL` промптов на ячейку класс × стиль подачи).

**Протокол**

1. Перебор в два этапа на одном seed: сначала общие гиперпараметры (один набор на все варианты), затем
   гиперпараметры самого loss — свои у каждого. Выбор по **pAUC** probe на val при FPR ∈ [0, `SELECTION_FPR_MAX`].
2. Полное обучение 5 вариантов × `SEEDS` с ранней остановкой по метрике мониторинга. Оценивается **лучший**
   чекпоинт по этой метрике — одно правило для всех.
3. Одинаковая оценка для всех моделей: эмбеддинг до головы → L2-нормализация → linear probe (C по val),
   kNN (k по val) и центроиды. `head` у вариантов с CE — отдельные строки.
4. Сравнение: среднее ± std по seed и парный бутстрэп разницы между вариантами.

**Метрики.** Классификация — AUROC, pAUC и TPR@FPR=1% (главные); FPR/TPR при порогах с val (1% и 5% FPR) и
при пороге, пересчитанном на самом наборе по квантилю, без меток. Индекс Юдена не используется. Геометрия до/после и по ходу обучения — isotropy, intrinsic dimensionality, kNN purity,
alignment/uniformity; при поломке фиксируются шаг и loss. Диагностика категорий вреда — FNR по категориям, kNN purity по
категории внутри harm и ARI kmeans, на hold-out (таксономия обучения) и на WildGuardMix (чужая таксономия).

## Конфигурация

`MAX_LENGTH` покрывает 95% текстов train. Длинные тексты обрезаются «голова + хвост»: шаблоны обёрток длинные,
а запрос может стоять и внутри, и в конце шаблона.

Эмбеддинги токенов (256k × 384, ~70% параметров модели) заморожены: при нынешнем объёме пар они почти не
получают сигнала.

**Батчи.** В каждом батче `CATEGORIES_PER_BATCH` категорий вреда по `HARM_PAIRS_PER_CATEGORY` harm-пар, остальное —
safe-пары. Категории выбираются с вероятностью ∝ √(число пар) — редкие чаще, чем по частоте. Сэмплер один и
тот же для всех вариантов и зависит только от seed. Число шагов — `len(пары) // BATCH_PAIRS × EPOCHS`.

**Гиперпараметры loss.** `DEFAULT_HPARAMS` — стартовые значения; перебор заменяет их лучшими по val. Варианты
наследуют найденное у «родителя» (`SWEEP_INHERITS`): например, `hier-supcon` берёт температуру из `supcon`.

## Train: пары v2

Пары собраны `build_pairs_v2` (см. `training_pairs_v2.md`): запросы обоих классов из AEGIS 2.0 train, обёртки
из HarmBench `HumanJailbreaks`, у каждого harm-текста — `primary_category` (первая категория без
`Needs Caution`, мелкие объединены в `Other`). Карта объединения категорий и шаблон «похож на jailbreak»
берутся из `manifest.json`, чтобы val и hold-out размечались теми же правилами.

`group` текста — идентификатор исходного промпта (`*_source_id`): у запроса и его обёрнутых версий он общий.
Он нужен kNN, чтобы один запрос не занимал сразу несколько мест среди соседей.

````python
PAIRS_DIR = PROJECT_DIR / PAIRS_SUBDIR
PAIR_FILES = (PAIRS_DIR / "pairs.jsonl", *([PAIRS_DIR / "code" / "pairs.jsonl"] if USE_CODE_PAIRS else []))
missing = [str(path) for path in PAIR_FILES if not path.is_file()]
if missing:
    raise FileNotFoundError(f"Нет файлов пар (сгенерируйте их локально build_pairs_v2): {missing}")

MANIFEST = json.loads((PAIRS_DIR / "manifest.json").read_text(encoding="utf-8"))
assert MANIFEST.get("schema_version") == 2, "ожидаются пары build_pairs_v2"
CATEGORY_MAP = MANIFEST["category_map"]
OTHER_CATEGORY = MANIFEST["other_category"]
DROPPED_CATEGORY = MANIFEST["dropped_category"]
JAILBREAK_RE = re.compile(MANIFEST["jailbreak_pattern"], flags=re.IGNORECASE)
# Персонажные джейлбрейки и промпт-инжекторные шаблоны попадают под один паттерн, но метка в них
# определяется по-разному, поэтому меряются они отдельно. Правило — из манифеста, чтобы train и
# оценка не разъехались молча.
PERSONA_JAILBREAK_RE = re.compile(MANIFEST["persona_jailbreak_pattern"], flags=re.IGNORECASE)
PERSONA_JAILBREAK, PROMPT_INJECTION_TEMPLATE = "persona_jailbreak", "prompt_injection_template"


def jailbreak_family(text: str) -> str | None:
    """Семейство jailbreak-подобного промпта, иначе None.

    Строгое разбиение среза: его состав не меняется. Персона важнее идиомы промпт-инжиниринга.
    """
    if not JAILBREAK_RE.search(text):
        return None
    return PERSONA_JAILBREAK if PERSONA_JAILBREAK_RE.search(text) else PROMPT_INJECTION_TEMPLATE
LABEL_IDS = {"safe": 0, "harm": 1}
CLASS_NAMES = ("safe", "harm")

TRAIN_PAIRS = pd.concat(
    [
        pd.read_json(path, lines=True, dtype={"anchor_source_id": str, "pair_source_id": str})
        for path in PAIR_FILES
    ],
    ignore_index=True,
)
for side in ("anchor", "pair"):
    TRAIN_PAIRS[f"{side}_y"] = TRAIN_PAIRS[f"{side}_label"].map(LABEL_IDS)
    TRAIN_PAIRS[f"{side}_category"] = TRAIN_PAIRS[f"{side}_category"].where(TRAIN_PAIRS[f"{side}_category"].notna(), None)
assert TRAIN_PAIRS[["anchor_y", "pair_y"]].notna().all().all(), "неизвестная метка в парах"

# категория harm-стороны пары; None — в паре только safe-тексты
TRAIN_PAIRS["harm_category"] = np.where(
    TRAIN_PAIRS["anchor_y"] == 1,
    TRAIN_PAIRS["anchor_category"],
    np.where(TRAIN_PAIRS["pair_y"] == 1, TRAIN_PAIRS["pair_category"], None),
)

TRAIN = (
    pd.concat([
        TRAIN_PAIRS[[f"{side}_text", f"{side}_y", f"{side}_category", f"{side}_source_id"]]
        .set_axis(["text", "label", "category", "group"], axis=1)
        for side in ("anchor", "pair")
    ])
    .drop_duplicates("text")
    .reset_index(drop=True)
)
TRAIN_TEXT_SET = set(TRAIN["text"])

HARM_CATEGORIES = sorted(TRAIN.loc[TRAIN["label"] == 1, "category"].dropna().unique())
CATEGORY_IDS = {category: index for index, category in enumerate(HARM_CATEGORIES)}

print(f"code-набор: {'да' if USE_CODE_PAIRS else 'нет'}; пар: {len(TRAIN_PAIRS)}, уникальных текстов: {len(TRAIN)}, "
      f"категорий вреда: {len(HARM_CATEGORIES)}")
display(TRAIN_PAIRS.groupby("pair_type").size().rename("pairs").to_frame())
display(
    TRAIN_PAIRS.dropna(subset=["harm_category"])
    .groupby("harm_category").size().rename("harm pairs").sort_values(ascending=False).to_frame()
)
````

## Val и hold-out: AEGIS 2.0 `validation` и `test`

Пары v2 собраны только из `train`-сплита AEGIS, и промпты val/test при сборке исключены. Здесь промпты
дедуплицируются, `REDACTED` и противоречивые метки выбрасываются. Категория harm-промпта считается по тем же
правилам, что при сборке; `jailbreak_like` — по тому же шаблону. Сами jailbreak-подобные промпты не
выбрасываются: на них отдельный срез метрик.

````python
AEGIS_PARQUET_URL = (
    "https://huggingface.co/api/datasets/nvidia/Aegis-AI-Content-Safety-Dataset-2.0/parquet/default/{split}/0.parquet"
)


def is_redacted(text: str) -> bool:
    return text.strip() == "REDACTED"


def primary_category(violated_categories) -> str:
    """Первая категория без Needs Caution, с тем же объединением мелких, что при сборке пар."""
    if not isinstance(violated_categories, str):
        return OTHER_CATEGORY
    categories = [category.strip() for category in violated_categories.split(",")]
    categories = [category for category in categories if category and category != DROPPED_CATEGORY]
    return CATEGORY_MAP.get(categories[0], OTHER_CATEGORY) if categories else OTHER_CATEGORY


def clean_eval_frame(frame: pd.DataFrame) -> pd.DataFrame:
    """Без пустых, REDACTED, дублей, противоречивых меток и текстов из train."""
    frame = frame[frame["text"].notna() & (frame["text"].str.strip() != "") & ~frame["text"].map(is_redacted)]
    frame = frame[~frame["text"].isin(TRAIN_TEXT_SET)]
    consistent = frame.groupby("text")["label"].transform("nunique") == 1
    return frame[consistent].drop_duplicates("text").reset_index(drop=True)


def load_aegis_split(split: str) -> pd.DataFrame:
    raw = pd.read_parquet(AEGIS_PARQUET_URL.format(split=split), columns=["prompt", "prompt_label", "violated_categories"])
    raw = raw[raw["prompt_label"].isin(["safe", "unsafe"])]
    harm = raw["prompt_label"] == "unsafe"
    text = raw["prompt"].astype(str).str.strip()
    return clean_eval_frame(pd.DataFrame({
        "text": text,
        "label": harm.astype(int),
        "category": [primary_category(value) if is_harm else None for value, is_harm in zip(raw["violated_categories"], harm)],
        "jailbreak_like": text.map(lambda value: JAILBREAK_RE.search(value) is not None),
        "jailbreak_family": text.map(jailbreak_family),
    }))


VAL = load_aegis_split("validation")
HOLDOUT = load_aegis_split("test")
````

## OOD: ToxicChat

`lmsys/toxic-chat`, конфигурация `toxicchat0124`, сплит test. Toxic-промптов в test ~360, поэтому в сэмпл идут

геометрия считается на сбалансированной подвыборке. Категорий вреда в ToxicChat нет.

# Общий код

## Модель

`GuardEncoder` — энкодер, masked mean pooling по последнему слою `h`, линейная голова safe/harm и projection
head `g` (`Linear → GELU → Linear`, `PROJECTION_DIM`). Классификационная голова обучается только в вариантах с
CE; projection head — только в контрастивных.

**Контрастивный loss считается на проекции `z = g(h)`, а probe, kNN, центроиды и все таблицы геометрии — на
`h`.** После обучения проекция выбрасывается. Так разрушение информации, которого требует контрастивный loss,
остаётся в выбрасываемом пространстве, а не в том, по которому оцениваются модели. Геометрия при этом
считается в обоих пространствах (`z · …`), чтобы отличить схлопывание `z` (ожидаемо) от схлопывания `h`.

````python
TOKENIZER = AutoTokenizer.from_pretrained(MODEL_NAME)

USE_AMP = DEVICE.type == "cuda"
# bfloat16 только на Ampere и новее; на T4 — float16 с GradScaler
AMP_DTYPE = torch.bfloat16 if USE_AMP and torch.cuda.get_device_capability()[0] >= 8 else torch.float16


def autocast():
    return torch.autocast(device_type=DEVICE.type, dtype=AMP_DTYPE, enabled=USE_AMP)


def tokenize(texts: Sequence[str], max_length: int = MAX_LENGTH) -> dict[str, torch.Tensor]:
    """Токенизация с обрезкой «голова + хвост» и паддингом до самого длинного текста."""
    body_length = max_length - 2
    head_length = int(body_length * TRUNCATION_HEAD_FRAC)
    rows = []
    for ids in TOKENIZER(list(texts), add_special_tokens=False)["input_ids"]:
        if len(ids) > body_length:
            ids = ids[:head_length] + ids[len(ids) - (body_length - head_length):]
        rows.append([TOKENIZER.cls_token_id, *ids, TOKENIZER.sep_token_id])

    width = max(len(ids) for ids in rows)
    input_ids = torch.full((len(rows), width), TOKENIZER.pad_token_id, dtype=torch.long)
    attention_mask = torch.zeros((len(rows), width), dtype=torch.long)
    for row, ids in enumerate(rows):
        input_ids[row, :len(ids)] = torch.tensor(ids)
        attention_mask[row, :len(ids)] = 1
    return {"input_ids": input_ids, "attention_mask": attention_mask}


class GuardEncoder(nn.Module):
    def __init__(self, model_name_or_path: str | Path, projection_dim: int | None = None,
                 projection_hidden: int | None = None) -> None:
        super().__init__()
        self.encoder = AutoModel.from_pretrained(model_name_or_path)
        hidden_size = self.encoder.config.hidden_size
        self.head = nn.Linear(hidden_size, len(CLASS_NAMES))
        self.projection = None if projection_dim is None else nn.Sequential(
            nn.Linear(hidden_size, projection_hidden or hidden_size),
            nn.GELU(),
            nn.Linear(projection_hidden or hidden_size, projection_dim),
        )

    def project(self, embeddings: torch.Tensor) -> torch.Tensor:
        """Пространство контрастивного loss. Без projection head — сам эмбеддинг."""
        return embeddings if self.projection is None else self.projection(embeddings)

    def forward(self, input_ids: torch.Tensor, attention_mask: torch.Tensor) -> tuple[torch.Tensor, torch.Tensor]:
        hidden_state = self.encoder(input_ids=input_ids, attention_mask=attention_mask).last_hidden_state
        mask = attention_mask.unsqueeze(-1).to(hidden_state.dtype)
        embeddings = (hidden_state * mask).sum(dim=1) / mask.sum(dim=1).clamp_min(1.0)
        return embeddings, self.head(embeddings)


@torch.inference_mode()
def embed_texts(model: GuardEncoder, texts: Sequence[str]) -> tuple[torch.Tensor, torch.Tensor]:
    """Эмбеддинги (n, d) и логиты головы (n, 2) на CPU. Тексты идут от длинных к коротким — меньше паддинга."""
    was_training = model.training
    model.eval()
    texts = list(texts)
    order = np.argsort([-len(text) for text in texts], kind="stable")

    embeddings, logits = [], []
    for start in range(0, len(texts), EMBED_BATCH_SIZE):
        batch = tokenize([texts[index] for index in order[start:start + EMBED_BATCH_SIZE]])
        with autocast():
            batch_embeddings, batch_logits = model(**{name: tensor.to(DEVICE) for name, tensor in batch.items()})
        embeddings.append(batch_embeddings.float().cpu())
        logits.append(batch_logits.float().cpu())

    inverse = np.empty_like(order)
    inverse[order] = np.arange(len(order))
    model.train(was_training)
    return torch.cat(embeddings)[inverse], torch.cat(logits)[inverse]


def embed_frames(model: GuardEncoder, frames: dict[str, pd.DataFrame]) -> dict[str, dict[str, torch.Tensor]]:
    outputs = {}
    for name, frame in frames.items():
        embeddings, logits = embed_texts(model, frame["text"])
        outputs[name] = {"embeddings": embeddings, "logits": logits}
    return outputs


def harm_score(logits: torch.Tensor) -> np.ndarray:
    """Лог-отношение шансов harm против safe."""
    return (logits[:, 1] - logits[:, 0]).numpy()


def save_model(model: GuardEncoder, path: Path, metadata: dict, save_head: bool) -> None:
    path.mkdir(parents=True, exist_ok=True)
    model.encoder.save_pretrained(path)
    TOKENIZER.save_pretrained(path)
    if save_head:  # у вариантов без CE голова не обучалась — её не сохраняем
        torch.save(model.head.state_dict(), path / "classifier_head.pt")
    if model.projection is not None:  # для справки: при оценке проекция не нужна, скореры читают h
        torch.save(model.projection.state_dict(), path / "projection_head.pt")
    (path / "rq3_training.json").write_text(json.dumps(metadata, ensure_ascii=False, indent=2, default=str), encoding="utf-8")


def wandb_enabled() -> bool:
    """Логирование включается заданием WANDB_PROJECT; без него блокнот работает как раньше."""
    return WANDB_PROJECT is not None


def wandb_login() -> None:
    """Логин заранее, до обучения: иначе wandb.init попросит ключ посреди длинной ячейки."""
    if not wandb_enabled():
        print("WANDB_PROJECT не задан — логирование в W&B выключено")
        return
    import wandb
    if "WANDB_API_KEY" not in os.environ:
        os.environ["WANDB_API_KEY"] = getpass.getpass("W&B API key (wandb.ai/authorize): ")
    wandb.login()
    print(f"W&B: вход выполнен, проект {WANDB_PROJECT}")


def wandb_start(name: str, job_type: str, group: str | None, config: dict, tags: Sequence[str] = ()):
    if not wandb_enabled():
        return None
    import wandb
    return wandb.init(project=WANDB_PROJECT, entity=WANDB_ENTITY, name=name, group=group,
                      job_type=job_type, tags=list(tags), config=config, reinit=True)


def wandb_log(run, values: dict, step: int | None = None) -> None:
    """Логируются только конечные числа: строки и NaN W&B не нужны."""
    if run is None:
        return
    clean = {name: float(value) for name, value in values.items()
             if isinstance(value, (int, float, np.floating, np.integer)) and np.isfinite(value)}
    if clean:
        run.log(clean, step=step)


def wandb_finish(run, summary: dict | None = None) -> None:
    if run is None:
        return
    for name, value in (summary or {}).items():
        run.summary[name] = value
    run.finish()


def free_memory() -> None:
    gc.collect()
    if DEVICE.type == "cuda":
        torch.cuda.empty_cache()
    elif DEVICE.type == "mps":
        torch.mps.empty_cache()


def upload_outputs(commit_message: str) -> None:
    """checkpoints/ и results/ — в приватный HF_CHECKPOINT_REPO; исключения в UPLOAD_IGNORE_PATTERNS."""
    if HF_CHECKPOINT_REPO is None:
        print("HF_CHECKPOINT_REPO не задан — выгрузка пропущена")
        return

    api = HfApi(token=os.environ["HF_TOKEN"])
    api.create_repo(HF_CHECKPOINT_REPO, repo_type="model", private=True, exist_ok=True)
    # exist_ok не меняет видимость уже существующего репозитория
    if not api.repo_info(HF_CHECKPOINT_REPO, repo_type="model").private:
        raise RuntimeError(f"{HF_CHECKPOINT_REPO} публичный — веса туда не выгружаются")

    api.upload_folder(
        repo_id=HF_CHECKPOINT_REPO,
        repo_type="model",
        folder_path=OUTPUT_DIR,
        allow_patterns=["checkpoints/*", "results/*"],
        ignore_patterns=UPLOAD_IGNORE_PATTERNS,
        commit_message=commit_message,
    )
    print(f"выгружено: https://huggingface.co/{HF_CHECKPOINT_REPO}")


wandb_login()
````

## Сэмплер батчей

Батч = `CATEGORIES_PER_BATCH` категорий вреда × `HARM_PAIRS_PER_CATEGORY` harm-пар + safe-пары до `BATCH_PAIRS`.
Harm-пара — пара, в которой есть harm-текст (`safe_harm_contrast` тоже: её safe-текст служит негативом).
Внутри батча пары не повторяются, между батчами выбираются с возвращением. Так у каждого harm-текста в батче
есть тексты той же категории, а партнёр по паре всегда рядом.

Для текста батч хранит:

* `categories` — индекс категории вреда (−1 у safe);
* `pair_index` — номер пары (общий у двух текстов пары);
* `source` — код исходного промпта (общий у запроса и его обёрнутых версий).

````python
def make_batch(rows: pd.DataFrame) -> dict:
    texts, labels, categories, sources = [], [], [], []
    for side in ("anchor", "pair"):
        texts += rows[f"{side}_text"].tolist()
        labels += rows[f"{side}_y"].tolist()
        categories += [
            CATEGORY_IDS.get(category, CATEGORY_IDS.get(OTHER_CATEGORY, -1)) if label == 1 else -1
            for category, label in zip(rows[f"{side}_category"], rows[f"{side}_y"])
        ]
        sources += rows[f"{side}_source_id"].tolist()
    return {
        "texts": texts,
        "labels": torch.tensor(labels, dtype=torch.long),
        "categories": torch.tensor(categories, dtype=torch.long),
        "pair_index": torch.arange(len(rows)).repeat(2),
        "source": torch.from_numpy(pd.factorize(pd.Series(sources))[0].astype(np.int64)),
    }


def iterate_category_batches(steps: int, seed: int):
    rng = np.random.default_rng(seed)
    harm_category = TRAIN_PAIRS["harm_category"]
    by_category = {
        category: np.flatnonzero((harm_category == category).to_numpy())
        for category in HARM_CATEGORIES
    }
    names = sorted(category for category, rows in by_category.items() if len(rows) >= HARM_PAIRS_PER_CATEGORY)
    weights = np.sqrt([len(by_category[name]) for name in names])
    weights /= weights.sum()
    safe_rows = np.flatnonzero(harm_category.isna().to_numpy())

    categories_per_batch = min(CATEGORIES_PER_BATCH, len(names))
    safe_per_batch = BATCH_PAIRS - categories_per_batch * HARM_PAIRS_PER_CATEGORY
    for _ in range(steps):
        chosen = rng.choice(len(names), size=categories_per_batch, replace=False, p=weights)
        rows = np.concatenate([
            *(rng.choice(by_category[names[index]], size=HARM_PAIRS_PER_CATEGORY, replace=False) for index in chosen),
            rng.choice(safe_rows, size=safe_per_batch, replace=False),
        ])
        yield make_batch(TRAIN_PAIRS.iloc[rows])
````

## Loss

* **CE** — cross-entropy головы по меткам текстов батча.
* **Взвешенный SupCon** — для текста i все остальные тексты батча участвуют в softmax по косинусному сходству
  (с температурой τ), а позитивы входят с весами:

| отношение текстов | `supcon` | `hier-supcon` |
|---|---|---|
| та же пара или тот же исходный промпт, тот же класс | 1 | 1 |
| оба harm, та же категория вреда | 1 | α |
| тот же класс, остальное (в том числе все safe) | 1 | β |
| разные классы | негатив | негатив |

Контрастив считается на проекции `z = g(h)`, CE — по логитам головы над `h`. В вариантах CE + контрастив
loss = CE + λ · SupCon.

````python
def weighted_supcon_loss(embeddings: torch.Tensor, positive_weights: torch.Tensor, temperature: float) -> torch.Tensor:
    z = F.normalize(embeddings.float(), dim=1)
    self_mask = torch.eye(len(z), dtype=torch.bool, device=z.device)
    log_prob = (z @ z.T / temperature).masked_fill(self_mask, float("-inf")).log_softmax(dim=1)
    weights = positive_weights.masked_fill(self_mask, 0.0)
    total = weights.sum(dim=1)
    loss = -(weights * log_prob.masked_fill(self_mask, 0.0)).sum(dim=1) / total.clamp_min(1e-8)
    return loss[total > 0].mean()


def positive_weights(batch: dict, kind: str, hparams: dict) -> torch.Tensor:
    labels = batch["labels"]
    same_class = labels[:, None] == labels[None, :]
    if kind == "class":
        return same_class.float()

    categories = batch["categories"]
    same_category = same_class & (labels[:, None] == 1) & (categories[:, None] == categories[None, :]) & (categories[:, None] >= 0)
    same_origin = same_class & (
        (batch["pair_index"][:, None] == batch["pair_index"][None, :]) | (batch["source"][:, None] == batch["source"][None, :])
    )
    weights = hparams["beta"] * same_class.float()
    weights = torch.where(same_category, torch.full_like(weights, hparams["alpha"]), weights)
    return torch.where(same_origin, torch.ones_like(weights), weights)


def compute_loss(variant: str, hparams: dict, projections: torch.Tensor, logits: torch.Tensor, batch: dict):
    """CE считается по логитам головы над эмбеддингом h, контрастив — над проекцией z = g(h)."""
    spec = VARIANTS[variant]
    parts = {}
    if spec["ce"]:
        parts["ce"] = F.cross_entropy(logits.float(), batch["labels"])
    if spec["contrastive"] is not None:
        weights = positive_weights(batch, spec["contrastive"], hparams)
        parts["contrastive"] = weighted_supcon_loss(projections, weights, hparams["temperature"])
    contrastive_weight = hparams["contrastive_weight"] if spec["ce"] else 1.0
    loss = parts.get("ce", 0.0) + contrastive_weight * parts.get("contrastive", 0.0)
    return loss, parts
````

## Метрики геометрии

Все метрики считаются на одной сбалансированной подвыборке, эмбеддинги — матрица (n, d):

* `mean cos` — **isotropy**: средний косинус по всем парам разных текстов, то есть матожидание для случайной
  пары. Около 1 — все векторы в узком конусе;
* `PCA-90 dim` — число главных компонент на 90% дисперсии; `PR dim` — participation ratio (Σλ)² / Σλ², число
  «примерно равноправных» осей; 
* `kNN purity` — доля из `KNN_K` ближайших по косинусу соседей с той же меткой; соседи ищутся внутри той же
  подвыборки, это диагностика геометрии, а не классификатор;
* `alignment (norm)` — E‖z₁ − z₂‖² по парам одного класса, **делённое на то же по всем парам**. Сырой
  alignment между моделями несравним: он в абсолютных расстояниях, а у моделей разная изотропия — frozen сидит
  в узком конусе, где все расстояния крошечные, и его сырой alignment минимален при полном отсутствии классовой
  структуры. Нормированный читается так: **1.0 — свои не ближе чужих**, ниже — классы плотнее облака, → 0 —
  class collapse. Учтите, что считается он по **бинарной** метке, то есть на уровне, которому `hier-supcon`
  намеренно даёт наименьший вес (β);
* `uniformity` — log E exp(−2‖z₁ − z₂‖²) по всем парам, z L2-нормированы. Меньше — равномернее покрыта сфера,
  0 — всё в одной точке.

**Геометрия сломалась**, если относительно шага 0 выполнено хоть одно: `mean cos` выше
`max(COLLAPSE_MEAN_COS, (1 + mean cos₀) / 2)`; `PR dim` ниже `COLLAPSE_DIM_FRACTION` от исходного;
`uniformity` выше `COLLAPSE_UNIFORMITY_FRACTION · uniformity₀`;

`category kNN purity` — то же, что `kNN purity`, но по категории вреда и только среди harm-текстов; `chance` —
её случайный уровень Σ p²_категории.

````python
GEOMETRY_COLUMNS = ("mean cos", "PCA-90 dim", "PR dim", "kNN purity", "alignment (norm)", "uniformity")


def normalized(x: torch.Tensor) -> torch.Tensor:
    return F.normalize(x.float(), dim=1)


def balanced_subsample(labels: np.ndarray, per_class: int = GEOMETRY_PER_CLASS, seed: int = SEED) -> np.ndarray:
    rng = np.random.default_rng(seed)
    size = min(per_class, *(int((labels == value).sum()) for value in np.unique(labels)))
    parts = [rng.permutation(np.flatnonzero(labels == value))[:size] for value in np.unique(labels)]
    return np.sort(np.concatenate(parts))


def get_components_at_threshold(explained_variance: torch.Tensor, var_threshold: float = VAR_THRESHOLD) -> int:
    cumulative_variance = explained_variance.cumsum(dim=0) / explained_variance.sum(dim=0)
    return int(torch.searchsorted(cumulative_variance, var_threshold).item()) + 1



def geometry_metrics(embeddings: torch.Tensor, labels: np.ndarray, k: int = KNN_K) -> dict[str, float]:
    x = embeddings.to(torch.float64)
    if not torch.isfinite(x).all():
        return {column: float("nan") for column in GEOMETRY_COLUMNS}

    y = torch.tensor(np.asarray(labels))
    n = len(x)
    z = F.normalize(x, dim=1)
    cos = z @ z.T
    off_diagonal = ~torch.eye(n, dtype=torch.bool)
    same_class = (y[:, None] == y[None, :]) & off_diagonal
    squared_distance = (2 - 2 * cos).clamp_min(0)

    centered = x - x.mean(dim=0)
    eigenvalues = torch.linalg.eigvalsh(centered.T @ centered / (n - 1)).flip(0).clamp_min(0)
    neighbours = cos.masked_fill(~off_diagonal, -2.0).topk(k, dim=1).indices

    return {
        "mean cos": cos[off_diagonal].mean().item(),
        "PCA-90 dim": float(get_components_at_threshold(eigenvalues)),
        "PR dim": (eigenvalues.sum() ** 2 / (eigenvalues ** 2).sum()).item(),
        "kNN purity": (y[neighbours] == y[:, None]).double().mean().item(),
        # Нормированный alignment: среднее расстояние внутри класса, делённое на среднее расстояние между
        # произвольными точками. Сырое значение между моделями несравнимо — у них разная изотропия, и у frozen
        # (конус, mean cos ≈ 0.95) все расстояния крошечные при полном отсутствии классовой структуры.
        # 1.0 — свои не ближе чужих; ниже — классы плотнее облака; → 0 — class collapse.
        "alignment (norm)": (squared_distance[same_class].mean() / squared_distance[off_diagonal].mean()).item(),
        "uniformity": torch.log(torch.exp(-2 * squared_distance[off_diagonal]).mean()).item(),
    }


def collapse_reasons(row: dict, reference: dict) -> list[str]:
    if not all(math.isfinite(row[key]) for key in GEOMETRY_COLUMNS):
        return ["эмбеддинги не конечны"]

    reasons = []
    mean_cos_limit = max(COLLAPSE_MEAN_COS, (1 + reference["mean cos"]) / 2)
    if row["mean cos"] > mean_cos_limit:
        reasons.append(f"mean cos {reference['mean cos']:.3f} → {row['mean cos']:.3f} (> {mean_cos_limit:.3f})")
    for key in ("PR dim",): 
        if row[key] < COLLAPSE_DIM_FRACTION * reference[key]:
            reasons.append(f"{key} {reference[key]:.1f} → {row[key]:.1f}")
    if row["uniformity"] > COLLAPSE_UNIFORMITY_FRACTION * reference["uniformity"]:
        reasons.append(f"uniformity {reference['uniformity']:.3f} → {row['uniformity']:.3f}")
    return reasons


def category_knn_purity(embeddings: torch.Tensor, categories: np.ndarray, k: int = KNN_K) -> tuple[float, float]:
    """kNN purity по категории среди harm-текстов и её случайный уровень."""
    z = normalized(embeddings).double()
    cos = (z @ z.T).fill_diagonal_(-2.0)
    neighbours = cos.topk(k, dim=1).indices
    codes = torch.from_numpy(pd.factorize(pd.Series(categories))[0])
    purity = (codes[neighbours] == codes[:, None]).double().mean().item()
    shares = pd.Series(categories).value_counts(normalize=True).to_numpy()
    return purity, float((shares ** 2).sum())
````

## Скореры и метрики классификации

У всех моделей одинаковые скореры поверх эмбеддинга до головы, после L2-нормализации:

* `probe` — логистическая регрессия; C выбирается из `PROBE_C_GRID` по **pAUC** на val для каждой модели отдельно;
* `knn` — k ближайших примеров train по косинусу, скор — доля harm, взвешенная по сходству. Соседи
  дедуплицируются по исходному промпту (`group`), чтобы запрос и его обёртки не голосовали несколько раз;
  k выбирается из `KNN_K_GRID` по **pAUC** на val;
* `centroid` — cos(z, μ_harm) − cos(z, μ_safe), центры по train, без параметров;
* `head` — лог-отношение шансов обученной головы, только у вариантов с CE. Для сравнения вариантов не
  используется: у `supcon` и `hier-supcon` обученной головы нет.

Отбор идёт по **pAUC** — нормированной площади под ROC на FPR ∈ [0, `SELECTION_FPR_MAX`]: именно там живёт
гардрейл, а сырой TPR@FPR=1% на val опирается на несколько промптов и для отбора слишком шумный.

Три разных вопроса — три группы метрик:

* `AUROC`, `pAUC` — качество **сортировки**, инвариантны к сдвигу шкалы скоров;
* `TPR@FPR=1%` — **оракул**: по ROC самого набора, то есть при идеально угаданном пороге;
* `FPR @ val N%`, `TPR @ val N%` — порог выбран на val при FPR ≤ N% (`THRESHOLD_FPR_GRID`) и перенесён без
  изменений, **как в деплое**. Отклонение фактического FPR от N% показывает, переносится ли калибровка.

Рекалибровка по квантилю скоров самого набора здесь не считается: она приближает FPR только при низкой доле
вреда в потоке, а все наборы оценки близки к сбалансированным — на hold-out верхний 1% скоров это сплошь harm,
и колонка вырождается в ноль. Вопрос рабочей точки в деплое надо мерить на пересэмпленном под реальную базовую
частоту наборе, отдельно от сравнения вариантов.

````python
def tpr_at_fpr(scores: np.ndarray, labels: np.ndarray, target: float = TARGET_FPR) -> float:
    fpr, tpr, _ = roc_curve(labels, scores)
    return float(tpr[fpr <= target].max())


def auroc(scores: np.ndarray, labels: np.ndarray) -> float:
    return float(roc_auc_score(labels, scores))


def partial_auroc(scores: np.ndarray, labels: np.ndarray, max_fpr: float = SELECTION_FPR_MAX) -> float:
    """Нормированный pAUC на FPR ∈ [0, max_fpr] (0.5 — случайно, 1 — идеально).

    По нему отбираются C, k и гиперпараметры: именно в этом углу ROC живёт гардрейл, а сырой
    TPR@FPR=1% на val опирается на ~6 safe-промптов и сам по себе слишком шумный для отбора.
    """
    return float(roc_auc_score(labels, scores, max_fpr=max_fpr))


def select_thresholds(scores: np.ndarray, labels: np.ndarray) -> dict[str, float]:
    """Пороги, дающие заданный FPR на val; переносятся на остальные наборы без изменений.

    Индекс Юдена не используется: он ставит каждую модель в свою точку (FPR 16–30%), из-за чего
    FNR между моделями несравним, и это не рабочая точка гардрейла.
    """
    fpr, _, thresholds = roc_curve(labels, scores)
    return {f"val {level:.0%}": float(thresholds[fpr <= level][-1]) for level in THRESHOLD_FPR_GRID}


def classification_metrics(scores: np.ndarray, labels: np.ndarray, thresholds: dict[str, float]) -> dict[str, float]:
    # Метрики, основанные на ROC, требуют достаточного числа промптов обоих классов.
    # При N негативов шаг сетки FPR равен 1/N: при 23 негативах условие «fpr <= 1%» оставляет
    # только точку fpr = 0, а при одном негативе бессмысленны и AUROC, и pAUC.
    # На таких срезах (например, persona jailbreaks: 1 safe против 18 harm) осмысленны только
    # пороговые метрики — сколько вредных модель флагает при общем пороге с val.
    negatives, positives = int((labels == 0).sum()), int((labels == 1).sum())
    ranking_is_meaningful = min(negatives, positives) >= MIN_CLASS_FOR_BOOTSTRAP
    row = {
        "AUROC": auroc(scores, labels) if ranking_is_meaningful else float("nan"),
        "pAUC": partial_auroc(scores, labels) if ranking_is_meaningful else float("nan"),
        "TPR@FPR=1%": tpr_at_fpr(scores, labels) if negatives >= MIN_NEGATIVES_FOR_TPR else float("nan"),
    }
    for name, threshold in thresholds.items():                      # порог перенесён с val
        flagged = scores >= threshold
        row[f"FPR @ {name}"] = float(flagged[labels == 0].mean())
        row[f"TPR @ {name}"] = float(flagged[labels == 1].mean())
    return row


def paired_bootstrap(metric, scores_a: np.ndarray, scores_b: np.ndarray, labels: np.ndarray,
                     rounds: int = BOOTSTRAP_ROUNDS) -> tuple[float, float, float]:
    """Разница metric(A) − metric(B) и её 95% интервал на одних и тех же бутстрэп-выборках."""
    rng = np.random.default_rng(SEED)
    negative, positive = np.flatnonzero(labels == 0), np.flatnonzero(labels == 1)
    differences = []
    for _ in range(rounds):
        index = np.concatenate([rng.choice(negative, len(negative)), rng.choice(positive, len(positive))])
        differences.append(metric(scores_a[index], labels[index]) - metric(scores_b[index], labels[index]))
    low, high = np.quantile(differences, [0.025, 0.975])
    return metric(scores_a, labels) - metric(scores_b, labels), float(low), float(high)


def fit_probe(train_x: torch.Tensor, train_y: np.ndarray, val_x: torch.Tensor, val_y: np.ndarray) -> dict:
    best = None
    for c in PROBE_C_GRID:
        probe = LogisticRegression(C=c, max_iter=5000).fit(normalized(train_x).numpy(), train_y)
        val_score = partial_auroc(probe.decision_function(normalized(val_x).numpy()), val_y)
        if best is None or val_score > best["val pAUC"]:
            best = {"model": probe, "C": c, "val pAUC": val_score}
    return best


def knn_scores(bank_x: torch.Tensor, bank_y: np.ndarray, bank_groups: np.ndarray, queries: torch.Tensor, k: int) -> np.ndarray:
    bank = normalized(bank_x)
    group_codes = pd.factorize(pd.Series(bank_groups))[0]
    candidates = min(len(bank), k * KNN_CANDIDATE_FACTOR)
    scores = np.empty(len(queries))
    for start in range(0, len(queries), 1024):
        similarities, indices = (normalized(queries[start:start + 1024]) @ bank.T).topk(candidates, dim=1)
        for row, (row_similarities, row_indices) in enumerate(zip(similarities.numpy(), indices.numpy())):
            # первое (самое близкое) вхождение каждого исходного промпта, затем k лучших
            _, first = np.unique(group_codes[row_indices], return_index=True)
            keep = np.sort(first)[:k]
            weights = np.clip(row_similarities[keep], 0.0, None) + 1e-6
            scores[start + row] = float((weights * bank_y[row_indices[keep]]).sum() / weights.sum())
    return scores


def centroid_scores(bank_x: torch.Tensor, bank_y: np.ndarray, queries: torch.Tensor) -> np.ndarray:
    bank = normalized(bank_x)
    harm_center = F.normalize(bank[bank_y == 1].mean(dim=0), dim=0)
    safe_center = F.normalize(bank[bank_y == 0].mean(dim=0), dim=0)
    z = normalized(queries)
    return (z @ harm_center - z @ safe_center).numpy()


def score_model(outputs: dict, with_head: bool) -> dict:
    """Скоры всех скореров на SCORED_DATASETS для одной модели и выбранные на val параметры."""
    train_x, train_y, groups = outputs["train"]["embeddings"], TRAIN["label"].to_numpy(), TRAIN["group"].to_numpy()
    val_y = VAL["label"].to_numpy()
    embeddings = {dataset: outputs[dataset]["embeddings"] for dataset in SCORED_DATASETS}
    results = {}

    probe = fit_probe(train_x, train_y, embeddings["val"], val_y)
    results["probe"] = {
        "params": {"C": probe["C"]},
        "model": probe["model"],
        "scores": {dataset: probe["model"].decision_function(normalized(x).numpy()) for dataset, x in embeddings.items()},
    }

    knn_val = {k: knn_scores(train_x, train_y, groups, embeddings["val"], k) for k in KNN_K_GRID}
    best_k = max(KNN_K_GRID, key=lambda k: partial_auroc(knn_val[k], val_y))
    results["knn"] = {
        "params": {"k": best_k},
        "scores": {
            dataset: knn_val[best_k] if dataset == "val" else knn_scores(train_x, train_y, groups, x, best_k)
            for dataset, x in embeddings.items()
        },
    }

    results["centroid"] = {
        "params": {},
        "scores": {dataset: centroid_scores(train_x, train_y, x) for dataset, x in embeddings.items()},
    }
    if with_head:
        results["head"] = {
            "params": {},
            "scores": {dataset: harm_score(outputs[dataset]["logits"]) for dataset in SCORED_DATASETS},
        }
    return results


def style_table(table: pd.DataFrame, caption: str):
    return (
        table.style
        .set_caption(caption)
        .format("{:.4f}", na_rep="—")
        .background_gradient(cmap="Blues", axis=0)
        .set_table_styles([
            {"selector": "th, td", "props": [
                ("border", "1px solid #9aa5b1"), ("padding", "6px 10px"), ("text-align", "center")]},
            {"selector": "caption", "props": [("font-weight", "bold"), ("text-align", "left")]},
        ])
    )


def mean_std_table(frame: pd.DataFrame, index: str, columns: Sequence[str], order: Sequence[str] | None = None) -> pd.DataFrame:
    """Среднее ± std по seed; у одиночных моделей (frozen) — только значение."""
    grouped = frame.groupby(index, sort=False)[list(columns)]
    mean, std = grouped.mean(), grouped.std()
    table = mean.map(lambda value: f"{value:.3f}") + std.map(lambda value: "" if pd.isna(value) else f" ± {value:.3f}")
    return table.reindex([name for name in (order or table.index) if name in table.index])
````

# Обучение

`train_variant` обучает один вариант с заданными гиперпараметрами и seed: линейный warmup + линейный decay,
AdamW, gradient clipping, батчи из `iterate_category_batches(steps, seed)`.

Каждые `EVAL_EVERY` шагов считается геометрия на `VAL_MONITOR` — в пространстве `h` и, у контрастивных
вариантов, дополнительно в `z`. При первом срабатывании правила поломки записываются шаг, эпоха, loss на этом
шаге, средний train loss с предыдущего замера и причины; параметры последнего здорового замера сохраняются
чекпоинтом `last_healthy_step*` — для справки.

**Ранняя остановка.** `EPOCHS` задаёт только верхнюю границу: обучение прерывается, когда `EARLY_STOP_METRIC`
на `VAL_MONITOR` не растёт `EARLY_STOP_PATIENCE` замеров подряд, и **оценивается лучший по ней чекпоинт, а не
финальный**. Остановка запрещена во время разогрева (`WARMUP_MAX_STEPS`), где метрика ещё не показательна.
Так каждый вариант сравнивается в своей лучшей точке, а не там, куда его привёл общий бюджет шагов.

Веса для seed из `SAVE_WEIGHTS_SEEDS` сохраняются в `checkpoints/<variant>/seed<seed>/final/`: энкодер и
токенизатор (`save_pretrained`), `classifier_head.pt` (только у вариантов с CE), `probe.joblib` (после оценки) и
`rq3_training.json` с гиперпараметрами и событием поломки.

````python
STEPS_PER_EPOCH = len(TRAIN_PAIRS) // BATCH_PAIRS
TOTAL_STEPS = EPOCHS * STEPS_PER_EPOCH if MAX_STEPS is None else MAX_STEPS

TRAINING_CONFIG = {
    "base_model": MODEL_NAME,
    "data": str(PAIRS_SUBDIR),
    "use_code_pairs": USE_CODE_PAIRS,
    "max_length": MAX_LENGTH,
    "truncation_head_frac": TRUNCATION_HEAD_FRAC,
    "batch_pairs": BATCH_PAIRS,
    "categories_per_batch": CATEGORIES_PER_BATCH,
    "harm_pairs_per_category": HARM_PAIRS_PER_CATEGORY,
    "head_lr": HEAD_LR,
    "weight_decay": WEIGHT_DECAY,
    "warmup_frac": WARMUP_FRAC,
    "warmup_max_steps": WARMUP_MAX_STEPS,
    "grad_clip": GRAD_CLIP,
    "freeze_token_embeddings": FREEZE_TOKEN_EMBEDDINGS,
    "projection_dim": PROJECTION_DIM,
    "projection_hidden": PROJECTION_HIDDEN,
    "early_stop_metric": EARLY_STOP_METRIC,
    "early_stop_patience": EARLY_STOP_PATIENCE,
    "pooling": "masked mean, last layer",
    "train_pairs": len(TRAIN_PAIRS),
}


def trainable_state(model: GuardEncoder) -> dict[str, torch.Tensor]:
    return {name: tensor.detach().to("cpu", copy=True) for name, tensor in model.named_parameters() if tensor.requires_grad}


def train_variant(variant: str, hparams: dict, seed: int, steps: int, frames: dict[str, pd.DataFrame], save: bool,
                  job_type: str = "train") -> dict:
    name = run_key(variant, seed)
    spec = VARIANTS[variant]
    wandb_run = wandb_start(
        name if job_type == "train" else f"sweep · {name} · {json.dumps(hparams, sort_keys=True)}",
        job_type, variant, {**hparams, "variant": variant, "seed": seed, "max_steps": steps, **TRAINING_CONFIG},
        tags=[job_type],
    )
    seed_everything(seed)
    # проекция нужна только там, где есть контрастивный loss; у чистого CE её нет
    projection_dim = PROJECTION_DIM if spec["contrastive"] is not None else None
    model = GuardEncoder(MODEL_NAME, projection_dim, PROJECTION_HIDDEN).to(DEVICE)
    if FREEZE_TOKEN_EMBEDDINGS:
        model.encoder.get_input_embeddings().requires_grad_(False)
    if not spec["ce"]:
        model.head.requires_grad_(False)
    if GRADIENT_CHECKPOINTING:
        # use_reentrant=False: иначе при замороженных эмбеддингах градиент не дойдёт до слоёв
        model.encoder.gradient_checkpointing_enable(gradient_checkpointing_kwargs={"use_reentrant": False})
    model.train()

    trainable = [tensor for tensor in model.parameters() if tensor.requires_grad]
    head_modules = [model.head] + ([model.projection] if model.projection is not None else [])
    head_ids = {id(tensor) for module in head_modules for tensor in module.parameters()}
    param_groups = [
        {"params": [tensor for tensor in trainable if id(tensor) not in head_ids], "lr": hparams["encoder_lr"]},
        {"params": [tensor for tensor in trainable if id(tensor) in head_ids], "lr": HEAD_LR},
    ]
    optimizer = torch.optim.AdamW([group for group in param_groups if group["params"]], weight_decay=WEIGHT_DECAY)
    warmup_steps = min(int(WARMUP_FRAC * steps), WARMUP_MAX_STEPS)
    scheduler = get_linear_schedule_with_warmup(optimizer, warmup_steps, steps)
    scaler = torch.amp.GradScaler("cuda", enabled=USE_AMP and AMP_DTYPE == torch.float16)

    monitor_labels = VAL_MONITOR["label"].to_numpy()
    step_log, history = [], []
    collapse, last_healthy = None, None
    best = {"score": -float("inf"), "step": 0, "params": None, "waits": 0}

    def record_collapse(step: int, reasons: list[str]) -> None:
        nonlocal collapse
        recent = [row["loss"] for row in step_log[-EVAL_EVERY:]]
        collapse = {
            "step": step,
            "epoch": step / STEPS_PER_EPOCH,
            "loss at step": step_log[-1]["loss"] if step_log else float("nan"),
            "mean train loss since previous eval": float(np.mean(recent)) if recent else float("nan"),
            "reasons": reasons,
            "metrics": {key: history[-1][key] for key in GEOMETRY_COLUMNS} if history else {},
        }
        print(f"[{name}] геометрия сломалась на шаге {step} (эпоха {collapse['epoch']:.2f}), "
              f"loss {collapse['loss at step']:.4f}: " + "; ".join(reasons))

    def evaluate(step: int) -> None:
        nonlocal last_healthy
        embeddings, logits = embed_texts(model, VAL_MONITOR["text"])
        recent = [row["loss"] for row in step_log[-EVAL_EVERY:]]
        row = {
            "variant": variant,
            "seed": seed,
            "step": step,
            "epoch": step / STEPS_PER_EPOCH,
            "train loss": float(np.mean(recent)) if recent else float("nan"),
            **geometry_metrics(embeddings, monitor_labels),
            "val AUROC (head)": auroc(harm_score(logits), monitor_labels) if spec["ce"] else float("nan"),
        }
        if model.projection is not None:
            # геометрия пространства контрастивного loss: схлопывание z ожидаемо и безвредно,
            # вопрос в том, доходит ли оно до h, по которому считаются все скореры
            with torch.inference_mode(), autocast():
                projected = model.project(embeddings.to(DEVICE)).float().cpu()
            row.update({f"z · {key}": value for key, value in geometry_metrics(projected, monitor_labels).items()})

        reasons = collapse_reasons(row, history[0]) if history else []
        row["collapsed"] = bool(reasons)
        history.append(row)
        if reasons and collapse is None:
            record_collapse(step, reasons)
        if collapse is None and save:
            last_healthy = {"step": step, "params": trainable_state(model)}

        score = row[EARLY_STOP_METRIC]
        if score > best["score"] + EARLY_STOP_MIN_DELTA:
            best.update(score=score, step=step, params=trainable_state(model), waits=0)
        else:
            best["waits"] += 1
        wandb_log(wandb_run, {f"eval/{key}": value for key, value in row.items()
                              if key not in ("variant", "seed", "step")}, step=step)
        print(f"[{name}] step {step:>4} · loss {row['train loss']:.4f} · mean cos {row['mean cos']:.3f} · "
              f"PR {row['PR dim']:.1f} · kNN purity {row['kNN purity']:.3f} · "
              f"best {EARLY_STOP_METRIC} {best['score']:.3f} @ {best['step']} (ждём {best['waits']})")

    evaluate(0)
    for step, batch in enumerate(iterate_category_batches(steps, seed), start=1):
        tokens = {key: tensor.to(DEVICE) for key, tensor in tokenize(batch["texts"]).items()}
        batch = {key: value.to(DEVICE) if torch.is_tensor(value) else value for key, value in batch.items()}
        with autocast():
            embeddings, logits = model(**tokens)
            projections = model.project(embeddings)
        loss, parts = compute_loss(variant, hparams, projections, logits, batch)

        optimizer.zero_grad(set_to_none=True)
        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(trainable, GRAD_CLIP)
        scaler.step(optimizer)
        scaler.update()
        scheduler.step()

        step_log.append({
            "variant": variant,
            "seed": seed,
            "step": step,
            "loss": loss.item(),
            **{part_name: part.item() for part_name, part in parts.items()},
            "lr": scheduler.get_last_lr()[0],
        })
        wandb_log(wandb_run, {f"train/{key}": value for key, value in step_log[-1].items()
                              if key not in ("variant", "seed", "step")}, step=step)
        if not math.isfinite(step_log[-1]["loss"]):
            if collapse is None:
                record_collapse(step, ["loss не конечен"])
            break
        if step % EVAL_EVERY == 0 or step == steps:
            evaluate(step)
        # во время разогрева метрика ещё не показательна — останавливаться по ней нельзя
        if step > warmup_steps and best["waits"] >= EARLY_STOP_PATIENCE:
            print(f"[{name}] ранняя остановка на шаге {step}: {EARLY_STOP_METRIC} не растёт "
                  f"{best['waits']} замеров, лучший шаг {best['step']}")
            break
        if STOP_ON_COLLAPSE and collapse is not None:
            break

    if best["params"] is not None and best["step"] != step_log[-1]["step"]:
        model.load_state_dict(best["params"], strict=False)   # оценивается лучший чекпоинт, не финальный
    outputs = embed_frames(model, frames)
    if save:
        metadata = {
            "variant": variant, "seed": seed, "loss": spec, "hparams": hparams,
            "steps": len(step_log), "best step": best["step"], **TRAINING_CONFIG, "collapse": collapse,
        }
        run_dir = CHECKPOINT_DIR / variant / f"seed{seed}"
        save_model(model, run_dir / "final", metadata, save_head=spec["ce"])
        if collapse is not None and last_healthy is not None and last_healthy["step"] > 0:
            model.load_state_dict(last_healthy["params"], strict=False)
            save_model(model, run_dir / f"last_healthy_step{last_healthy['step']}", {**metadata, "steps": last_healthy["step"]}, save_head=spec["ce"])

    wandb_finish(wandb_run, {
        "best step": best["step"],
        f"best {EARLY_STOP_METRIC}": best["score"],
        "steps": len(step_log),
        "collapsed": collapse is not None,
        **({"collapse/step": collapse["step"], "collapse/reasons": "; ".join(collapse["reasons"])} if collapse else {}),
    })

    # освобождаем GPU до следующего запуска: веса и состояния оптимизатора больше не нужны
    model.to("cpu")
    optimizer.state.clear()
    free_memory()
    return {
        "variant": variant,
        "seed": seed,
        "hparams": hparams,
        "best step": best["step"],
        "best score": best["score"],
        "history": pd.DataFrame(history),
        "steps": pd.DataFrame(step_log),
        "collapse": collapse,
        "outputs": outputs,
    }
````

## Перебор гиперпараметров

Короткие запуски (`SWEEP_STEPS` шагов) на первом seed. Качество конфигурации — **pAUC** linear probe на val:
probe учится на сбалансированной подвыборке train (`SWEEP_PROBE_TRAIN_TEXTS`). Hold-out и OOD-наборы в переборе
не участвуют.

**Два этапа.** Сначала `SHARED_GRID` на `SHARED_SWEEP_VARIANT` подбирает общие гиперпараметры
(`SHARED_HPARAMS`) — и они применяются ко **всем** вариантам. Затем `SWEEP_GRIDS` крутит гиперпараметры самого
loss, свои у каждого варианта, поверх общих. После перебора `assert` проверяет, что общие гиперпараметры у всех
вариантов совпали: иначе сравнение «варианты отличаются только loss» молча перестаёт выполняться.

Выбирается лучшая конфигурация среди тех, где геометрия не сломалась; если сломалась везде — лучшая из всех,
и это печатается предупреждением, чтобы заведомо поломанный режим не уехал в основные запуски незамеченным.

Варианты идут в порядке `VARIANTS`, поэтому «родитель» из `SWEEP_INHERITS` к моменту перебора потомка уже выбран.

````python
assert RUN_SOURCE == "train", (
    f"RUN_SOURCE = {RUN_SOURCE!r} — перебор относится к пути «обучение», пропустите его"
)

SWEEP_SEED = SEEDS[0]
SWEEP_TRAIN = TRAIN.iloc[balanced_subsample(TRAIN["label"].to_numpy(), SWEEP_PROBE_TRAIN_TEXTS // 2)].reset_index(drop=True)
SWEEP_FRAMES = {"sweep-train": SWEEP_TRAIN, "val": VAL}
SWEEP_PATH = RESULTS_DIR / "sweep.pt"

sweep_cache = torch.load(SWEEP_PATH, weights_only=False) if SWEEP_PATH.exists() else {}
BEST_HPARAMS, sweep_rows = {}, []


def sweep_config(variant: str, hparams: dict, override: dict, stage: str) -> dict:
    """Короткий прогон одной конфигурации; результат кэшируется по отпечатку данных и настроек."""
    key = fingerprint(DATA_FINGERPRINT, variant, hparams, SWEEP_SEED, SWEEP_STEPS, SWEEP_PROBE_TRAIN_TEXTS, TRAINING_CONFIG)
    if key not in sweep_cache:
        print(f"=== sweep [{stage}] {variant}: {override}")
        run = train_variant(variant, hparams, SWEEP_SEED, SWEEP_STEPS, SWEEP_FRAMES, save=False, job_type="sweep")
        probe = fit_probe(
            run["outputs"]["sweep-train"]["embeddings"], SWEEP_TRAIN["label"].to_numpy(),
            run["outputs"]["val"]["embeddings"], VAL["label"].to_numpy(),
        )
        sweep_cache[key] = {
            "stage": stage,
            "variant": variant,
            "config": json.dumps(override),
            **{f"hp {name}": value for name, value in hparams.items()},
            "val pAUC (probe)": probe["val pAUC"],
            "probe C": probe["C"],
            "collapsed": run["collapse"] is not None,
            "collapse step": run["collapse"]["step"] if run["collapse"] else None,
            "final PR dim": float(run["history"]["PR dim"].iloc[-1]),
            "hparams": hparams,
        }
        torch.save(sweep_cache, SWEEP_PATH)
    return sweep_cache[key]


def pick_best(rows: list[dict]) -> dict:
    """Лучшая по val pAUC среди несломавшихся; если геометрия сломалась везде — лучшая из всех.

    Вторая ветка — молчаливый откат, и в прошлом эксперименте она выбрала supcon с τ = 0.1,
    сломавшейся на шаге 25: вариант сравнивался в заведомо поломанном режиме. Теперь откат
    печатается, чтобы такая конфигурация не уезжала в основные запуски незамеченной.
    """
    healthy = [row for row in rows if not row["collapsed"]]
    if not healthy:
        variant = rows[0]["variant"]
        print(f"  ВНИМАНИЕ [{variant}]: геометрия сломалась во всех {len(rows)} конфигурациях — "
              f"выбор идёт среди поломанных, результаты варианта интерпретировать с оговоркой")
        healthy = rows
    return max(healthy, key=lambda row: row["val pAUC (probe)"])


# Этап 1: общие гиперпараметры — один подбор на все варианты.
SHARED_BEST = {}
if RUN_SWEEP and SHARED_GRID:
    shared_rows = [sweep_config(SHARED_SWEEP_VARIANT, {**DEFAULT_HPARAMS, **override}, override, "shared")
                   for override in SHARED_GRID]
    SHARED_BEST = {name: pick_best(shared_rows)["hparams"][name] for name in SHARED_HPARAMS}
    sweep_rows += shared_rows
    print(f"общие для всех вариантов: {SHARED_BEST}\n")

# Этап 2: гиперпараметры loss — свои у каждого варианта, поверх общих.
for variant in VARIANTS:
    base = {**DEFAULT_HPARAMS, **SHARED_BEST, **BEST_HPARAMS.get(SWEEP_INHERITS.get(variant), {})}
    if not RUN_SWEEP:
        BEST_HPARAMS[variant] = base
        continue
    variant_rows = [sweep_config(variant, {**base, **override}, override, variant) for override in SWEEP_GRIDS[variant]]
    BEST_HPARAMS[variant] = pick_best(variant_rows)["hparams"]
    sweep_rows += variant_rows

# Страховка от повторения прошлой ошибки: общие гиперпараметры обязаны совпасть у всех вариантов.
for name in SHARED_HPARAMS:
    values = {variant: hparams[name] for variant, hparams in BEST_HPARAMS.items()}
    assert len(set(values.values())) == 1, f"{name} разошёлся между вариантами: {values}"

(RESULTS_DIR / "best_hparams.json").write_text(json.dumps(BEST_HPARAMS, indent=2), encoding="utf-8")
if sweep_rows:
    SWEEP = pd.DataFrame(sweep_rows).drop(columns="hparams")
    SWEEP.to_csv(RESULTS_DIR / "sweep.csv", index=False)
    display(SWEEP[["stage", "variant", "config", "val pAUC (probe)", "probe C",
                   "collapsed", "collapse step", "final PR dim"]])
display(pd.DataFrame(BEST_HPARAMS).T)
````

## Основные запуски

5 вариантов × `SEEDS` на полное число шагов `TOTAL_STEPS` с выбранными гиперпараметрами. Внутри одного seed все
варианты видят одинаковые батчи и одинаковую инициализацию головы, поэтому их можно сравнивать попарно.

````python
assert RUN_SOURCE == "train", (
    f"RUN_SOURCE = {RUN_SOURCE!r} — обучение пропускается, модели уже подняты из чекпоинтов"
)

RUNS_PATH = RESULTS_DIR / "runs.pt"
RUNS = torch.load(RUNS_PATH, weights_only=False) if RUNS_PATH.exists() else {}

for seed in SEEDS:
    for variant in VARIANTS:
        key = run_key(variant, seed)
        run_fingerprint = fingerprint(DATA_FINGERPRINT, variant, BEST_HPARAMS[variant], seed, TOTAL_STEPS, TRAINING_CONFIG)
        if RUNS.get(key, {}).get("fingerprint") == run_fingerprint:
            print(f"{key}: из кэша")
            continue
        print(f"=== {key}: {BEST_HPARAMS[variant]}, {TOTAL_STEPS} шагов")
        run = train_variant(variant, BEST_HPARAMS[variant], seed, TOTAL_STEPS, EVAL_FRAMES, save=seed in SAVE_WEIGHTS_SEEDS)
        RUNS[key] = {**run, "fingerprint": run_fingerprint}
        torch.save(RUNS, RUNS_PATH)

RUN_KEYS = [run_key(variant, seed) for seed in SEEDS for variant in VARIANTS]
HISTORY = pd.concat([RUNS[key]["history"] for key in RUN_KEYS], ignore_index=True)
STEP_LOG = pd.concat([RUNS[key]["steps"] for key in RUN_KEYS], ignore_index=True)
COLLAPSES = {key: RUNS[key]["collapse"] for key in RUN_KEYS}

(RESULTS_DIR / "geometry_collapse.json").write_text(json.dumps(COLLAPSES, ensure_ascii=False, indent=2), encoding="utf-8")
HISTORY.to_csv(RESULTS_DIR / "training_geometry_history.csv", index=False)
STEP_LOG.to_csv(RESULTS_DIR / "training_steps.csv", index=False)
````

## Кривые обучения

Линия — среднее по seed, полоса — от минимума до максимума. Loss — скользящее среднее по 10 шагам; шкалы CE и
SupCon разные, поэтому loss сравнивается по форме кривой. Остальные панели — замеры на `VAL_MONITOR`; тонкие
вертикальные линии — шаги, на которых у отдельных seed сработало правило поломки.

````python
# Кривые существуют только после обучения: в чекпоинте истории нет, а в старых runs.pt
# колонки записаны до переименования alignment -> alignment (norm).
assert RUN_SOURCE == "train" and not HISTORY.empty, (
    "кривых обучения нет: они пишутся только при RUN_SOURCE = 'train'"
)

from matplotlib.ticker import MaxNLocator

VARIANT_COLORS = {
    "ce": "#2a78d6",
    "supcon": "#eb6834",
    "ce+supcon": "#1baf7a",
    "hier-supcon": "#eda100",
    "ce+hier-supcon": "#e87ba4",
}
SURFACE = "#fcfcfb"
GRIDLINE = "#e1e0d9"
BASELINE = "#c3c2b7"
INK_SECONDARY = "#52514e"
INK_MUTED = "#898781"

CURVE_PANELS = ("loss", "val AUROC (head)", "kNN purity", "mean cos", "PR dim",
                "PCA-90 dim", "alignment (norm)", "uniformity")


def plot_training_curves(history: pd.DataFrame, step_log: pd.DataFrame, collapses: dict):
    figure, axes = plt.subplots(3, 3, figsize=(14, 10.5), squeeze=False, facecolor=SURFACE)
    smoothed = step_log.assign(
        loss=step_log.groupby(["variant", "seed"])["loss"].transform(lambda series: series.rolling(10, min_periods=1).mean())
    )

    for axis, panel in zip(axes.flat, CURVE_PANELS):
        source = smoothed if panel == "loss" else history
        for variant, color in VARIANT_COLORS.items():
            rows = source[source["variant"] == variant]
            if rows.empty or rows[panel].isna().all():
                continue
            stats = rows.groupby("step")[panel].agg(["mean", "min", "max"])
            axis.fill_between(stats.index, stats["min"], stats["max"], color=color, alpha=0.1, linewidth=0)
            if panel == "loss":
                axis.plot(stats.index, stats["mean"], color=color, linewidth=2, solid_capstyle="round", label=variant)
            else:
                axis.plot(stats.index, stats["mean"], color=color, linewidth=2, solid_capstyle="round",
                          solid_joinstyle="round", marker="o", markersize=6,
                          markeredgecolor=SURFACE, markeredgewidth=1.5, label=variant)
            for key, collapse in collapses.items():
                if collapse and key.startswith(f"{variant}@"):
                    axis.axvline(collapse["step"], color=color, linewidth=1, alpha=0.5)

        axis.set_facecolor(SURFACE)
        axis.set_title(panel, fontsize=10, color=INK_SECONDARY, loc="left")
        axis.grid(axis="y", color=GRIDLINE, linewidth=0.75)
        axis.set_axisbelow(True)
        axis.tick_params(colors=INK_MUTED, length=0)
        for side, spine in axis.spines.items():
            spine.set_visible(side == "bottom")
            spine.set_color(BASELINE)
    for axis in axes.flat:
        axis.xaxis.set_major_locator(MaxNLocator(integer=True))
    for axis in axes[-1]:
        axis.set_xlabel("шаг", color=INK_MUTED)

    handles, labels = axes[0, 0].get_legend_handles_labels()
    figure.suptitle("Обучение: loss и геометрия на val (среднее и разброс по seed)", x=0.01, y=0.995, ha="left", color=INK_SECONDARY)
    figure.legend(handles, labels, loc="upper left", bbox_to_anchor=(0.005, 0.965), ncols=len(labels),
                  frameon=False, labelcolor=INK_SECONDARY)
    figure.tight_layout(rect=(0, 0, 1, 0.94))
    plt.close(figure)
    return figure


TRAINING_FIGURE = plot_training_curves(HISTORY, STEP_LOG, COLLAPSES)
TRAINING_FIGURE.savefig(RESULTS_DIR / "training_curves.png", dpi=150, facecolor=SURFACE)
display(TRAINING_FIGURE)
````

тоже на val. Оценивается лучший чекпоинт запуска; его probe сохраняется рядом с весами (`probe.joblib`) для seed
из `SAVE_WEIGHTS_SEEDS`.

````python
MODEL_OUTPUTS = {"frozen": FROZEN_OUTPUTS, **{key: RUNS[key]["outputs"] for key in RUN_KEYS}}
MODEL_INFO = {
    "frozen": {"variant": "frozen", "seed": None},
    **{key: {"variant": RUNS[key]["variant"], "seed": RUNS[key]["seed"]} for key in RUN_KEYS},
}
MODEL_ORDER = ["frozen", *VARIANTS]
VAL_LABELS = VAL["label"].to_numpy()

SCORES, classification_rows, scorer_params = {}, [], []
for model_name, outputs in MODEL_OUTPUTS.items():
    info = MODEL_INFO[model_name]
    with_head = info["variant"] != "frozen" and VARIANTS[info["variant"]]["ce"]
    SCORES[model_name] = score_model(outputs, with_head)

    for scorer, result in SCORES[model_name].items():
        scorer_params.append({"model": model_name, **info, "scorer": scorer, **result["params"]})
        thresholds = select_thresholds(result["scores"]["val"], VAL_LABELS)
        for set_name, (dataset, mask) in CLASSIFICATION_SETS.items():
            classification_rows.append({
                "model": model_name,
                **info,
                "scorer": scorer,
                "dataset": set_name,
                **classification_metrics(
                    result["scores"][dataset][mask], EVAL_FRAMES[dataset]["label"].to_numpy()[mask], thresholds
                ),
            })

    final_dir = CHECKPOINT_DIR / str(info["variant"]) / f"seed{info['seed']}" / "final"
    if info["seed"] in SAVE_WEIGHTS_SEEDS and final_dir.exists():
        joblib.dump(SCORES[model_name]["probe"]["model"], final_dir / "probe.joblib")
    print(f"{model_name}: готово")

CLASSIFICATION = pd.DataFrame(classification_rows)
CLASSIFICATION.to_csv(RESULTS_DIR / "classification_runs.csv", index=False)
pd.DataFrame(scorer_params).to_csv(RESULTS_DIR / "scorer_params.csv", index=False)

# Порог Юдена убран: он ставил каждую модель в свою точку и делал FNR несравнимым между моделями.
# «val N%» — порог выбран на val при FPR <= N% и перенесён без изменений.
MAIN_METRICS = ["AUROC", "pAUC", "TPR@FPR=1%",
                "FPR @ val 1%", "TPR @ val 1%", "TPR @ val 5%"]
(
    CLASSIFICATION.groupby(["variant", "scorer", "dataset"], sort=False)[MAIN_METRICS]
    .agg(["mean", "std"])
    .to_csv(RESULTS_DIR / "classification_summary.csv")
)
````

Одни и те же сбалансированные подвыборки для всех моделей, среднее ± std по seed. Вторая таблица — разница
средних с frozen: рост `kNN purity` — классы чище; рост `mean cos`, падение размерностей и рост `uniformity` —

`alignment (norm)` нормирован на среднее расстояние между произвольными точками, поэтому сравним между
моделями: **1.0 — свои не ближе чужих** (так выглядит frozen), ниже — классы плотнее облака, → 0 — class

Низкий `alignment (norm)` сам по себе **не признак качества**: минимума достигает именно class collapse. Плюс
считается он по **бинарной** метке, то есть на уровне, которому `hier-supcon` намеренно даёт наименьший вес (β),
— его рост у иерархических вариантов это работа целевой функции, а не деградация. Читать надо вместе с
`PR dim` и `uniformity`.

## Категории вреда

Проверка, сохраняется ли в пространстве структура по категориям вреда. Считается на `CATEGORY_DATASETS`:

* **hold-out** — таксономия AEGIS, та самая, на которой училась иерархия (уровень α). Категория промпта
  размечается по тем же правилам, что в train — карта объединения берётся из `manifest.json`;
* **WildGuardMix** — **чужая** таксономия (`subcategory`), которой модель не видела. Если структура держится
  и здесь, это перенос геометрии типов вреда, а не запоминание наших меток.

Обе метрики работают с любым разбиением harm-множества, поэтому отображать одну таксономию в другую не нужно;
сравниваются модели между собой на одном наборе, а не абсолютные значения между наборами. Показываются
категории, в которых не меньше `MIN_CATEGORY_EVAL` harm-промптов.

* **FNR по категориям** — доля пропущенных harm-промптов категории при пороге `CATEGORY_FNR_THRESHOLD` с val,
  основной скорер. Порог **один и тот же для всех моделей**, иначе FNR между ними несравним: при пороге,
  подобранном каждой модели отдельно, различия в FNR — это различия порогов, а не моделей. Общий TPR при этом
  скрывает, какие именно виды вреда модель пропускает;
* **category kNN purity** — среди harm-промптов доля соседей той же категории; `chance` — случайный уровень.
  Если после обучения она падает к `chance`, обучение стирает различия между видами вреда, даже когда бинарные
  метрики растут;
* **ARI kmeans** — совпадение кластеров kmeans с разметкой: `class` — 2 кластера на сбалансированной подвыборке,
  `category` — по кластеру на категорию среди harm. Это диагностика геометрии, не метрика классификации.

````python
CATEGORY_FNR_THRESHOLD = f"val {THRESHOLD_FPR_GRID[-1]:.0%}"   # общий для всех моделей порог
# hold-out — таксономия, на которой училась иерархия; WildGuard — чужая, проверка переноса структуры.
CATEGORY_DATASETS = [name for name in ("hold-out", "WildGuard")
                     if name in SCORED_DATASETS and EVAL_FRAMES[name]["category"].notna().any()]

fnr_rows, diagnostic_rows, eval_categories = [], [], {}
for dataset in CATEGORY_DATASETS:
    frame = EVAL_FRAMES[dataset]
    labels = frame["label"].to_numpy()
    categories = frame["category"].to_numpy(dtype=object)
    counts = frame.loc[frame["label"] == 1, "category"].value_counts()
    eval_categories[dataset] = [c for c, n in counts.items() if n >= MIN_CATEGORY_EVAL]
    harm_rows = np.flatnonzero((labels == 1) & frame["category"].isin(eval_categories[dataset]).to_numpy())
    display(counts.rename(f"{dataset} harm prompts").to_frame())

    for model_name, outputs in MODEL_OUTPUTS.items():
        info = MODEL_INFO[model_name]
        scores = SCORES[model_name][PRIMARY_SCORER]["scores"]
        # один и тот же FPR на val у всех моделей — иначе FNR по категориям между ними несравним
        missed = scores[dataset] < select_thresholds(scores["val"], VAL_LABELS)[CATEGORY_FNR_THRESHOLD]
        fnr_rows.append({
            "dataset": dataset, "model": model_name, **info,
            **{category: float(missed[(labels == 1) & (categories == category)].mean())
               for category in eval_categories[dataset]},
        })

        embeddings = outputs[dataset]["embeddings"]
        purity, chance = category_knn_purity(embeddings[harm_rows], categories[harm_rows])
        index = GEOMETRY_INDEX[dataset]
        ari_class = adjusted_rand_score(
            labels[index],
            KMeans(n_clusters=2, n_init=10, random_state=SEED).fit_predict(normalized(embeddings[index]).numpy()),
        )
        ari_category = float("nan")
        if len(eval_categories[dataset]) >= 2:
            ari_category = adjusted_rand_score(
                categories[harm_rows],
                KMeans(n_clusters=len(eval_categories[dataset]), n_init=10, random_state=SEED)
                .fit_predict(normalized(embeddings[harm_rows]).numpy()),
            )
        diagnostic_rows.append({
            "dataset": dataset, "model": model_name, **info,
            "category kNN purity": purity, "chance": chance,
            "ARI kmeans · class": ari_class, "ARI kmeans · category": ari_category,
        })

CATEGORY_FNR = pd.DataFrame(fnr_rows)
CATEGORY_DIAGNOSTICS = pd.DataFrame(diagnostic_rows)
CATEGORY_FNR.to_csv(RESULTS_DIR / "category_fnr.csv", index=False)
CATEGORY_DIAGNOSTICS.to_csv(RESULTS_DIR / "category_diagnostics.csv", index=False)

for dataset in CATEGORY_DATASETS:
    rows = CATEGORY_FNR[CATEGORY_FNR["dataset"] == dataset]
    means = rows.groupby("variant")[eval_categories[dataset]].mean()
    display(style_table(
        means.reindex([variant for variant in MODEL_ORDER if variant in means.index]).T,
        f"FNR по категориям на {dataset} · {PRIMARY_SCORER} · порог {CATEGORY_FNR_THRESHOLD} (среднее по seed)",
    ))
    print(f"{dataset} · структура категорий")
    display(mean_std_table(
        CATEGORY_DIAGNOSTICS[CATEGORY_DIAGNOSTICS["dataset"] == dataset], "variant",
        ["category kNN purity", "chance", "ARI kmeans · class", "ARI kmeans · category"], MODEL_ORDER,
    ))
````

````python
from sklearn.manifold import TSNE

BEST_SEED = SAVE_WEIGHTS_SEEDS[0]
CATEGORY_PANELS = 6                # категорий на фигуру: больше — нечитаемо
HIGHLIGHT_COLOR = "#2a78d6"        # один валидированный оттенок; идентичность несёт панель, а не цвет
CONTEXT_COLOR = "#c9c9c4"


def plot_projections(projections, colorings, title, point_size=7.0):
    """Строки — модели, колонки — раскраски."""
    figure, axes = plt.subplots(len(projections), len(colorings),
                                figsize=(5.2 * len(colorings), 4.6 * len(projections)), squeeze=False)
    palette = plt.rcParams["axes.prop_cycle"].by_key()["color"]

    for row, (model_name, projection) in enumerate(projections.items()):
        for column, (coloring_name, labels) in enumerate(colorings.items()):
            axis = axes[row][column]
            for color_number, value in enumerate(sorted(set(labels.tolist()))):
                mask = labels == value
                axis.scatter(projection[mask, 0], projection[mask, 1], s=point_size, alpha=0.55,
                             linewidths=0, color=palette[color_number % len(palette)], label=str(value))
            axis.set_title(f"{model_name} · t-SNE · {coloring_name}", fontsize=10)
            axis.set_xticks([]); axis.set_yticks([]); axis.grid(alpha=0.2)
            if row == 0:
                axis.legend(markerscale=2.5, fontsize=8)

    figure.suptitle(title)
    figure.tight_layout()
    plt.close(figure)
    return figure


def plot_category_highlights(projections, harm_mask, categories, title, panels=CATEGORY_PANELS):
    """Строки — модели, колонки — категории вреда: своя категория поверх остальных harm-точек.

    Small multiples вместо раскраски в 16 цветов: на scatter различимы три-четыре оттенка,
    дальше цвета перестают читаться, и структура категорий не видна именно там, где её проверяют.
    """
    harm_categories = categories[harm_mask]
    top = pd.Series(harm_categories).value_counts().index[:panels].tolist()
    figure, axes = plt.subplots(len(projections), len(top),
                                figsize=(2.9 * len(top), 3.0 * len(projections)), squeeze=False)

    for row, (model_name, projection) in enumerate(projections.items()):
        points = projection[harm_mask]
        for column, category in enumerate(top):
            axis, mask = axes[row][column], harm_categories == category
            axis.scatter(points[:, 0], points[:, 1], s=5, alpha=0.30, linewidths=0, color=CONTEXT_COLOR)
            axis.scatter(points[mask, 0], points[mask, 1], s=10, alpha=0.9, linewidths=0, color=HIGHLIGHT_COLOR)
            axis.set_xticks([]); axis.set_yticks([]); axis.grid(alpha=0.2)
            if row == 0:
                axis.set_title(f"{category}\n(n = {int(mask.sum())})", fontsize=8)
            if column == 0:
                axis.set_ylabel(model_name, fontsize=8)

    figure.suptitle(title)
    figure.tight_layout()
    plt.close(figure)
    return figure


def tsne(embeddings):
    return TSNE(n_components=2, perplexity=30.0, init="pca",
                learning_rate="auto", metric = "cosine", random_state=SEED).fit_transform(normalized(embeddings).numpy())


# Эмбеддинги уже посчитаны при обучении (WildGuard входит в EVAL_FRAMES) — переэмбеддить и грузить
# чекпоинты не нужно. Варианты заданы явно, без отбора по метрикам: нужно конкретное сравнение
# «class-SupCon стирает категории» против «иерархия их сохраняет».
TSNE_VARIANTS = ("supcon", "hier-supcon")
TSNE_WITH_FROZEN = True            # frozen как точка отсчёта; False — оставить только два варианта

TSNE_MODELS = {
    **({"mmBERT-small (frozen)": "frozen"} if TSNE_WITH_FROZEN else {}),
    **{f"mmBERT-small · {variant}": run_key(variant, BEST_SEED) for variant in TSNE_VARIANTS},
}

PROJECTIONS = {dataset: {} for dataset in ("WildGuard", "hold-out")}
for label, model_name in TSNE_MODELS.items():
    for dataset in PROJECTIONS:
        PROJECTIONS[dataset][label] = tsne(MODEL_OUTPUTS[model_name][dataset]["embeddings"])
    print(f"{label}: готово")

WILDGUARD_FIGURE = plot_projections(
    PROJECTIONS["WildGuard"],
    {"класс": np.where(WILDGUARD["label"] == 1, "harmful", "benign"), "стиль": WILDGUARD["style"].to_numpy()},
    "WildGuardMix · t-SNE",
)
WILDGUARD_FIGURE.savefig(RESULTS_DIR / "tsne_wildguardmix.png", dpi=120)
display(WILDGUARD_FIGURE)

# Главная фигура: сохраняется ли близость текстов одного типа вреда внутри harm.
# Сначала таксономия, на которой училась иерархия (AEGIS hold-out), затем чужая (WildGuardMix) — перенос структуры.
for dataset, title in (("hold-out", "AEGIS hold-out · категории вреда внутри harm"),
                       ("WildGuard", "WildGuardMix · подкатегории вреда внутри harm (чужая таксономия)")):
    frame = EVAL_FRAMES[dataset]
    figure = plot_category_highlights(PROJECTIONS[dataset], (frame["label"] == 1).to_numpy(),
                                      frame["category"].to_numpy(dtype=object), title)
    figure.savefig(RESULTS_DIR / f"tsne_categories_{dataset.replace(' ', '_')}.png", dpi=120)
    display(figure)
````

## Adversarial vs vanilla на WildGuardMix

Внешняя проверка механизма, который задают пары v2. `jailbreak_variant` и `benign_twin` используют одни и те
же 114 шаблонов HarmBench на обоих классах, поэтому в train обвязка не несёт информации о метке, а партнёр по
паре — позитив с весом 1. Оптимум loss — убрать подпространство обвязки из представления.

Если это так, то на промптах, где обвязка информативна, дообученные модели должны терять больше, чем frozen.
Величина для проверки — разрыв между стилями подачи внутри одной модели:

```
gap = AUROC(vanilla) − AUROC(adversarial)
предсказание: gap у дообученных > gap у frozen
```

Разрыв считается внутри модели, поэтому из него уходит и сложность самих подмножеств, и общее качество модели;
сравнение разрывов между моделями — разность разностей. CI — стратифицированный парный бутстрэп по четырём
ячейкам (стиль × класс).

````python
# Скоры уже посчитаны основным скорером: WildGuard входит в EVAL_FRAMES, probe обучен на train,
# C выбран по val. Набор остаётся полностью hold-out.
GAP_MODELS = ["frozen", *VARIANTS]
GAP_SEED = SAVE_WEIGHTS_SEEDS[0]

wildguard_labels = WILDGUARD["label"].to_numpy()
is_adversarial = (WILDGUARD["style"] == "adversarial").to_numpy()
wildguard_tokens = np.array([len(ids) for ids in TOKENIZER(WILDGUARD["text"].tolist(), add_special_tokens=False)["input_ids"]])

display(
    pd.DataFrame({"tokens": wildguard_tokens, "label": wildguard_labels, "style": WILDGUARD["style"]})
    .groupby(["label", "style"])["tokens"].describe(percentiles=[0.5, 0.95])[["count", "50%", "95%"]]
)

# vanilla×safe, vanilla×harm, adversarial×safe, adversarial×harm
GAP_STRATA = [np.flatnonzero((is_adversarial == adversarial) & (wildguard_labels == label))
              for adversarial in (False, True) for label in (0, 1)]


def model_key(variant: str) -> str:
    return "frozen" if variant == "frozen" else run_key(variant, GAP_SEED)


def style_auroc(scores: np.ndarray) -> tuple[float, float, float]:
    """AUROC на vanilla, на adversarial и разрыв между ними."""
    vanilla = auroc(scores[~is_adversarial], wildguard_labels[~is_adversarial])
    adversarial = auroc(scores[is_adversarial], wildguard_labels[is_adversarial])
    return vanilla, adversarial, vanilla - adversarial


def paired_gap_bootstrap(scores_a: np.ndarray, scores_b: np.ndarray, rounds: int = BOOTSTRAP_ROUNDS):
    """Разность разностей (gap_a − gap_b) и её 95% интервал: перевыборка внутри четырёх страт."""
    rng = np.random.default_rng(SEED)
    differences = []
    for _ in range(rounds):
        picked = [rng.choice(stratum, len(stratum)) for stratum in GAP_STRATA]
        vanilla, adversarial = np.concatenate(picked[:2]), np.concatenate(picked[2:])
        gaps = [auroc(scores[vanilla], wildguard_labels[vanilla]) - auroc(scores[adversarial], wildguard_labels[adversarial])
                for scores in (scores_a, scores_b)]
        differences.append(gaps[0] - gaps[1])
    low, high = np.quantile(differences, [0.025, 0.975])
    return float(np.mean(differences)), float(low), float(high)


WILDGUARD_SCORES, gap_rows = {}, []
for variant in GAP_MODELS:
    scores = WILDGUARD_SCORES[variant] = SCORES[model_key(variant)][PRIMARY_SCORER]["scores"]["WildGuard"]
    vanilla, adversarial, gap = style_auroc(scores)
    row = {"model": variant, "AUROC vanilla": vanilla, "AUROC adversarial": adversarial, "gap": gap}
    # санитарная проверка: не объясняется ли скор просто длиной текста
    for name, mask in (("vanilla", ~is_adversarial), ("adversarial", is_adversarial)):
        row[f"corr(score, tokens) · {name}"] = float(
            pd.Series(scores[mask]).corr(pd.Series(wildguard_tokens[mask]), method="spearman")
        )
    gap_rows.append(row)
    print(f"{variant}: vanilla {vanilla:.4f} · adversarial {adversarial:.4f} · gap {gap:.4f}")

WILDGUARD_GAP = pd.DataFrame(gap_rows).set_index("model").reindex(GAP_MODELS)
WILDGUARD_GAP.to_csv(RESULTS_DIR / "wildguard_style_gap.csv")
display(style_table(WILDGUARD_GAP, f"WildGuardMix · AUROC по стилю подачи · {PRIMARY_SCORER} · seed {GAP_SEED}"))
````

### Разность разностей с frozen

````python
gap_diff_rows = []
for model_name in GAP_MODELS:
    if model_name == "frozen":
        continue
    difference, low, high = paired_gap_bootstrap(WILDGUARD_SCORES[model_name], WILDGUARD_SCORES["frozen"])
    gap_diff_rows.append({
        "comparison": f"{model_name} − frozen",
        "Δgap": difference,
        "CI low": low,
        "CI high": high,
        "gap значимо больше": bool(low > 0),
        "gap значимо меньше": bool(high < 0),
    })

WILDGUARD_GAP_DIFF = pd.DataFrame(gap_diff_rows).set_index("comparison")
WILDGUARD_GAP_DIFF.to_csv(RESULTS_DIR / "wildguard_style_gap_diff.csv")
display(
    WILDGUARD_GAP_DIFF.style
    .format("{:.4f}", subset=["Δgap", "CI low", "CI high"])
    .set_caption("Разность разностей с frozen · Δgap > 0 — дообучение теряет на adversarial сильнее, чем frozen")
)

confirmed = int(WILDGUARD_GAP_DIFF["gap значимо больше"].sum())
refuted = int(WILDGUARD_GAP_DIFF["gap значимо меньше"].sum())
print(
    f"\nпредсказание «gap обученных > gap frozen»: подтверждено у {confirmed} из {len(WILDGUARD_GAP_DIFF)} вариантов, "
    f"опровергнуто у {refuted}"
)
````

## Retrieval: `h` против `z`

Projection head `g` выбрасывается при оценке, и все скореры выше работают на `h`. Но retrieval — это
ближайшие соседи по косинусу, а `z = g(h)` — ровно то пространство, где контрастивный loss напрямую
организовывал косинусную близость. Здесь проверяется, не лучше ли `z` для поиска.

Три вопроса:

| метрика | что проверяет |
|---|---|
| **бинарный kNN** — AUROC, pAUC, TPR@FPR=1% | годится ли пространство для kNN-гардрейла. Банк — train, `k` по pAUC на val, как у скорера `knn` |
| **category purity** | соседи того же вида вреда — на hold-out и на чужой таксономии WildGuard |
| **twin hit@k** | есть ли среди k ближайших обёрнутая версия того же запроса или его парафраза |

````python
RETRIEVAL_SEED = SAVE_WEIGHTS_SEEDS[0]
RETRIEVAL_DATASETS = ("hold-out", "ToxicChat", "WildGuard")
TWIN_K = (1, 5, 10)


def load_projection(variant: str, seed: int) -> nn.Module | None:
    """Projection head из чекпоинта; форма восстанавливается по сохранённым весам. У ce её нет."""
    path = CHECKPOINT_DIR / variant / f"seed{seed}" / "final" / "projection_head.pt"
    if not path.exists():
        return None
    state = torch.load(path, map_location="cpu")
    hidden, inputs = state["0.weight"].shape
    projection = nn.Sequential(nn.Linear(inputs, hidden), nn.GELU(), nn.Linear(hidden, state["2.weight"].shape[0]))
    projection.load_state_dict(state)
    return projection.eval()


@torch.inference_mode()
def project_outputs(outputs: dict, projection: nn.Module) -> dict:
    """z = g(h) для всех наборов. h в MODEL_OUTPUTS ненормированный — так же, как на входе g при обучении."""
    return {name: {"embeddings": projection(frame["embeddings"].float())} for name, frame in outputs.items()}


SPACES = {("frozen", "h"): MODEL_OUTPUTS["frozen"]}
for variant in VARIANTS:
    key = run_key(variant, RETRIEVAL_SEED)
    if key not in MODEL_OUTPUTS:
        continue
    SPACES[(variant, "h")] = MODEL_OUTPUTS[key]
    projection = load_projection(variant, RETRIEVAL_SEED)
    if projection is not None:
        SPACES[(variant, "z")] = project_outputs(MODEL_OUTPUTS[key], projection)

print("пространства:", ", ".join(f"{variant} · {space}" for variant, space in SPACES))
````

````python
def space_order(frame: pd.DataFrame) -> pd.DataFrame:
    """frozen первым, дальше варианты в порядке VARIANTS, у каждого h перед z."""
    order = [(v, s) for v in MODEL_ORDER for s in ("h", "z") if (v, s) in set(zip(frame["variant"], frame["space"]))]
    return frame.set_index(["variant", "space"]).loc[order]


# бинарный kNN: банк train, k по pAUC на val, как у скорера knn
train_y, train_groups, val_y = TRAIN["label"].to_numpy(), TRAIN["group"].to_numpy(), VAL["label"].to_numpy()
knn_rows = []
for (variant, space), outputs in SPACES.items():
    bank = outputs["train"]["embeddings"]
    on_val = {k: knn_scores(bank, train_y, train_groups, outputs["val"]["embeddings"], k) for k in KNN_K_GRID}
    best_k = max(KNN_K_GRID, key=lambda k: partial_auroc(on_val[k], val_y))
    for dataset in RETRIEVAL_DATASETS:
        labels = EVAL_FRAMES[dataset]["label"].to_numpy()
        scores = knn_scores(bank, train_y, train_groups, outputs[dataset]["embeddings"], best_k)
        knn_rows.append({"variant": variant, "space": space, "dataset": dataset, "k": best_k,
                         "AUROC": auroc(scores, labels), "pAUC": partial_auroc(scores, labels),
                         "TPR@FPR=1%": tpr_at_fpr(scores, labels)})
RETRIEVAL_KNN = pd.DataFrame(knn_rows)
RETRIEVAL_KNN.to_csv(RESULTS_DIR / "retrieval_knn.csv", index=False)
for metric in ("AUROC", "TPR@FPR=1%"):
    table = space_order(RETRIEVAL_KNN.pivot_table(index=["variant", "space"], columns="dataset",
                                                  values=metric).reset_index())[list(RETRIEVAL_DATASETS)]
    display(style_table(table, f"Бинарный kNN · {metric} · h против z"))

# --- категории вреда среди соседей
category_rows = []
for (variant, space), outputs in SPACES.items():
    for dataset in CATEGORY_DATASETS:
        frame = EVAL_FRAMES[dataset]
        categories = frame["category"].to_numpy(dtype=object)
        harm = np.flatnonzero((frame["label"] == 1).to_numpy() & frame["category"].isin(eval_categories[dataset]).to_numpy())
        purity, chance = category_knn_purity(outputs[dataset]["embeddings"][harm], categories[harm])
        category_rows.append({"variant": variant, "space": space, "dataset": dataset,
                              "category purity": purity, "chance": chance})
RETRIEVAL_CATEGORIES = pd.DataFrame(category_rows)
RETRIEVAL_CATEGORIES.to_csv(RESULTS_DIR / "retrieval_categories.csv", index=False)
table = space_order(RETRIEVAL_CATEGORIES.pivot_table(index=["variant", "space"], columns="dataset",
                                                     values="category purity").reset_index())[list(CATEGORY_DATASETS)]
display(style_table(table, "Category purity среди harm · h против z"))
print("случайный уровень:", RETRIEVAL_CATEGORIES.groupby("dataset")["chance"].first().round(3).to_dict())

# --- twin hit@k: находится ли обёртка или парафраза того же запроса среди ближайших (in-sample)
text_index = {text: index for index, text in enumerate(TRAIN["text"])}


def collect_twins(pair_types: tuple[str, ...], symmetric: bool) -> dict[int, set[int]]:
    pairs = TRAIN_PAIRS[TRAIN_PAIRS["pair_type"].isin(pair_types)]
    twins: dict[int, set[int]] = {}
    for anchor, partner in zip(pairs["anchor_text"], pairs["pair_text"]):
        a, p = text_index[anchor], text_index[partner]
        twins.setdefault(a, set()).add(p)
        if symmetric:
            twins.setdefault(p, set()).add(a)
    return twins


TWINS = {
    "обёртка": collect_twins(("jailbreak_variant", "benign_twin"), symmetric=False),   # сырой запрос → его обёртки
}


def twin_hit_at_k(embeddings: torch.Tensor, twins: dict[int, set[int]], ks: tuple[int, ...]) -> dict[int, float]:
    """Доля запросов, у которых хотя бы один «близнец» среди k ближайших; сам запрос из поиска исключён."""
    bank, queries = normalized(embeddings), np.fromiter(twins, dtype=np.int64)
    hits = {k: 0 for k in ks}
    for start in range(0, len(queries), 1024):
        batch = queries[start:start + 1024]
        similarity = bank[torch.from_numpy(batch)] @ bank.T
        similarity[torch.arange(len(batch)), torch.from_numpy(batch)] = -2.0
        top = similarity.topk(max(ks), dim=1).indices.numpy()
        for row, query in zip(top, batch):
            for k in ks:
                hits[k] += bool(twins[query].intersection(row[:k]))
    return {k: hits[k] / len(queries) for k in ks}


twin_rows = []
for (variant, space), outputs in SPACES.items():
    for relation, twins in TWINS.items():
        for k, value in twin_hit_at_k(outputs["train"]["embeddings"], twins, TWIN_K).items():
            twin_rows.append({"variant": variant, "space": space, "relation": relation, "k": k, "hit@k": value})
RETRIEVAL_TWINS = pd.DataFrame(twin_rows)
RETRIEVAL_TWINS.to_csv(RESULTS_DIR / "retrieval_twins.csv", index=False)
for relation in TWINS:
    rows = RETRIEVAL_TWINS[RETRIEVAL_TWINS["relation"] == relation]
    table = space_order(rows.pivot_table(index=["variant", "space"], columns="k", values="hit@k").reset_index())
    table.columns = [f"hit@{k}" for k in table.columns]
    display(style_table(table, f"Twin hit@k · {relation} · train (in-sample) · запросов: {len(TWINS[relation])}"))
````
